```python
df['DateTime'] = pd.to_datetime(df['DateTime'], format="%d-%m-%Y %H:%M", errors='coerce')
```

```python
# Fix datetime
df['DateTime'] = pd.to_datetime(df['DateTime'], errors='coerce')
df = df.dropna(subset=['DateTime'])
```

```python
df['Date'] = df['DateTime'].dt.date
df['Hour'] = df['DateTime'].dt.hour
df['Day'] = df['DateTime'].dt.day_name()
df['Month'] = df['DateTime'].dt.month_name()
```

```python
# Outlier computation via IQR method
Q1 = df['Vehicles'].quantile(0.25)
Q3 = df['Vehicles'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Vehicles'] < lower_bound) | (df['Vehicles'] > upper_bound)]
print(f"Statistical boundaries: Lower={lower_bound}, Upper={upper_bound}")
print(f"Total number of outlier rows detected: {len(outliers)}")
```
